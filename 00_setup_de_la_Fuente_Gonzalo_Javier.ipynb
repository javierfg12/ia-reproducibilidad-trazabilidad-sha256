## Objetivo
Preparar un entorno reproducible y comprobar la integridad de los artefactos mediante SHA-256.

## Procedimiento
1. Identificar el archivo que se desea verificar.
2. Calcular su huella SHA-256.
3. Compararla con el valor registrado en `registro_hash.json`.
4. Documentar cualquier discrepancia antes de continuar.

```python
from pathlib import Path
import hashlib
import json

ARCHIVO = Path("00_setup_de_la_Fuente_Gonzalo_Javier.ipynb")
REGISTRO = Path("registro_hash.json")

def sha256_archivo(ruta: Path) -> str:
    """Devuelve el SHA-256 del contenido binario de un archivo."""
    digest = hashlib.sha256()
    with ruta.open("rb") as archivo:
        for bloque in iter(lambda: archivo.read(65536), b""):
            digest.update(bloque)
    return digest.hexdigest()

if not ARCHIVO.is_file():
    raise FileNotFoundError(f"No se encuentra el archivo: {ARCHIVO}")
if not REGISTRO.is_file():
    raise FileNotFoundError(f"No se encuentra el registro: {REGISTRO}")

registro = json.loads(REGISTRO.read_text(encoding="utf-8"))
hash_calculado = sha256_archivo(ARCHIVO)
hash_registrado = registro.get("sha256", "")

print(f"Archivo: {ARCHIVO.name}")
print(f"SHA-256 calculado: {hash_calculado}")
print(f"SHA-256 registrado: {hash_registrado}")
print("Integridad correcta" if hash_calculado == hash_registrado else "ALERTA: el hash no coincide")
```

## Criterios de reproducibilidad
- Mantener el archivo de registro junto al artefacto.
- Recalcular el hash después de cada modificación del archivo.
- No considerar una coincidencia de hash como prueba de calidad funcional: solo acredita que el contenido coincide con el registrado.
- Registrar fecha, archivo y resultado de cada verificación.
